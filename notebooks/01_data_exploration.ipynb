```python
from pathlib import Path
import pandas as pd

DATA_DIR = Path("../data/raw/premier_league")

files = sorted(DATA_DIR.glob("E0_*.csv"))

print(f"Liczba plików: {len(files)}")

for file in files:
    print(file.name)
```

```python
dfs = []

for file in files:
    if file.name == "E0_0405.csv":
        df = pd.read_csv(
            file,
            encoding="cp1252",
            engine="python",
            usecols=range(57)
        )
    else:
        df = pd.read_csv(
            file,
            encoding="cp1252"
        )

    df["Season"] = file.stem.replace("E0_", "")
    dfs.append(df)

data = pd.concat(dfs, ignore_index=True)

print(f"Liczba meczów: {len(data)}")
print(f"Liczba kolumn: {len(data.columns)}")
```
